## What could have been done better

### Notebook 01 — EDA

1. **"Normally distributed" claim was too strong** — we only checked skewness. To claim normality we'd need kurtosis, Q-Q plots, or a statistical test. The honest claim is "approximately symmetric."

2. **The 12-feature distribution chart was weak** — first 12 features alphabetically gives a misleading sample. Should have plotted the **top 12 most discriminative features** (by mean difference) to actually show what useful features look like.

3. **Outlier statement was unjustified** — we said "consistent with real financial data" but the features are anonymized; we have no idea what they represent.

4. **We missed a critical Santander-specific insight** — the dataset has a special property: features have many repeated values (mean 97,907 unique values out of 200,000 rows). This was THE key insight that top Kaggle solutions exploited via **frequency encoding** (creating a feature that counts how often each value appears). We never explored this.

### Notebook 02 — Feature Engineering

5. **Aggregate features were probably wasted effort** — their correlation with target was ~0.02. We kept them anyway for educational purposes, but a real ML project would have dropped them after testing.

6. **Only one feature engineering technique tried** — we didn't explore frequency encoding, feature binning, or interactions. For 200 features with weak individual signal, more techniques are usually worth testing.

### Notebook 03 — Modeling

7. **No cross-validation** — we used a single 80/20 split. For a class project, **5-fold cross-validation** would give more reliable AUC estimates and protect against lucky splits.

8. **LightGBM was barely tuned** — we used arbitrary hyperparameters (1000 trees, learning_rate 0.05, num_leaves 31). No grid search, no early stopping, no validation curve. This is likely why we fell short of 0.90 AUC.

9. **No early stopping** — LightGBM supports early stopping based on validation AUC. We trained 1000 trees blindly when fewer or more might have been better.

10. **Confusion matrix was disappointing** (class 1 precision 0.44) — but we never actually compared to the baseline LR's confusion matrix to see if LightGBM is actually better on the imbalanced class, or just better at ranking.

## What to prioritize fixing

If time is limited, the **three highest-impact improvements** are:

1. **Add frequency encoding** in notebook 02 — most likely to push AUC past 0.90
2. **Add 5-fold cross-validation** in notebook 03 — more reliable evaluation
3. **Tune LightGBM hyperparameters** with early stopping — better model quality
